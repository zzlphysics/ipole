```python
hfp = h5py.File("/root/Projects/ipole/output/riaf/job1/files_20241017/0c3nme8oil.h5",'r')    
dx = hfp['header']['camera']['dx'][()]
dsource = hfp['header']['dsource'][()]
print(dx, dsource)
```

```python
file_list = os.listdir("/root/Projects/ipole/output/riaf/job1/res_fake_gauss")

for file in file_list:
    print(file)
    # file的文件名形如-0.002056_930918.760736_321677203260.021362_0.209172_5147660463.141061_0.752199_-0.999999_59.000349.png
    # 提取文件名中的数字，以_为分隔符，提取出数字，忽略最后的.png
    nums = file.split("_")
    nums = [float(num) for num in nums[:-1]]
    bh_mass = nums[4]

    G_CONS = 6.6742e-8
    C_CONS = 2.99792458e10
    M_SUN = 1.989e33
    PC = 3.085678e18
    lunit = G_CONS * bh_mass * M_SUN / (C_CONS * C_CONS)
    MUAS_PER_RAD = 2.06265e11
    fovx_dsource = 160
    dsource_pc = 16.8e6
    dsource = dsource_pc * PC
    fov_to_d = dsource / lunit / MUAS_PER_RAD
    dx = fovx_dsource * fov_to_d
    print(dx, dsource)
    # 计算视场大小（以微弧秒为单位）
    fov_muas = dx / dsource * lunit * 2.06265e11
    # 读取灰度图像为二维数组
    img = Image.open("/root/Projects/ipole/output/riaf/job1/res_fake_gauss/" + file)
    img = img.convert('L')
    img = np.array(img)
    # print(img.shape)
    nx = img.shape[0]
    rad2muas = np.pi/180/3600/1000000
    sigma = 20 / (2 * np.sqrt(2 * np.log(2))) / ((dx*lunit/nx)/(dsource)/rad2muas)

    Ic = gaussian_filter(img, sigma)

    Ic_normalized = (Ic / Ic.max() * 255).astype(np.uint8)  # 归一化到 0-255 范围
    image_Ic = Image.fromarray(Ic_normalized, mode='L')  # 创建灰度图像

    # # 保存图像为 320x320 像素，避免插值
    image_Ic.save("/root/Projects/ipole/output/riaf/job1/res_fake_gauss_20/" +file.replace(".png", "_Ic.png"))
    
```
