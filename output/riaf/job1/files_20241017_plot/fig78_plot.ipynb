```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
plt.rcParams.update({
    "font.family": "Times New Roman",
    "font.size": 8,        # 默认字体大小
    "axes.labelsize": 8,   # 轴标签字体大小
    "legend.fontsize": 8,  # 图例字体大小
    "mathtext.fontset": "stix",  # 数学字体设置为 STIX，这个与 Times New Roman 很接近
})
```

```python
def read_test_data(file_path):
    """
    读取测试数据文件，并将数据转换为numpy数组
    
    参数:
        file_path (str): 测试数据文件的路径
        
    返回:
        tuple: 包含三个numpy数组 (sample_indices, predictions, labels)
    """
    # 读取数据
    data = pd.read_csv(file_path, sep=',', skipinitialspace=True)
    
    # 提取列数据
    sample_indices = data.iloc[:, 0].values
    predictions = data.iloc[:, 1].values
    labels = data.iloc[:, 2].values
    
    return sample_indices, predictions, labels
```

```python
file_path = "/root/Projects/ipole/output/riaf/job1/files_20241017_plot/DDPM/test_result_data/a_test_data.txt"
indices, preds, labels = read_test_data(file_path)

predictions = np.array(preds)
label_list = np.array(labels)

plt.figure(figsize=(8, 4))
plt.scatter(np.arange(len(predictions)), predictions, color='#015876', s=10, label='Predictions')  # s 控制点的大小
plt.scatter(np.arange(len(label_list)), label_list, color='#FDB137', s=10, label='Labels')  # s 控制点的大小
for i in range(len(predictions)):
    plt.plot([i, i], [predictions[i], label_list[i]], color='#C1D1D1', linestyle='-', linewidth=1)
plt.xlabel('sample')
plt.ylabel('number')
plt.ylim(-2.5, 2.5)
plt.legend()
plt.show()
```
