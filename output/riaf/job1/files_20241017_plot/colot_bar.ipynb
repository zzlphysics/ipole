```python
def colorbar(mappable):
  """ the way matplotlib colorbar should have been implemented """
  from mpl_toolkits.axes_grid1 import make_axes_locatable
  ax = mappable.axes
  fig = ax.figure
  divider = make_axes_locatable(ax)
  cax = divider.append_axes("right", size="5%", pad=0.05)
  return fig.colorbar(mappable, cax=cax)
```

```python
ddpm_dir = "/root/Projects/ipole/output/riaf/job1/files_20241017_plot/fig_add_colorbar/fig6_DDPM/"
grrt_dir = "/root/Projects/ipole/output/riaf/job1/files_20241017_plot/fig_add_colorbar/fig6_GRRT/"
origin_dir = "/root/Projects/ipole/output/riaf/job1/files_20241017_outofrange_20250301_1/"
output_dir = "/root/Projects/ipole/output/riaf/job1/files_20241017_plot/fig_add_colorbar/fig6_DDPM_colorbar/"
os.makedirs(output_dir, exist_ok=True)

file_name = "0.392066_9821628.851327_152369125510.644653_0.034368_5392217229.831393_0.223765_-1.000000_220.743639_L.png"
hdf5_file_name = "79biu5gu1m.h5"

# file_name = "0.604742_53203.163368_1734529894146.155518_0.063061_14145789491.095959_0.006038_-1.000000_3.301814_L.png"
# hdf5_file_name = "a5ash5w9cz.h5"

# file_name = "0.620040_729108.844775_934138982809.583130_0.683237_1811393521.104505_0.269101_-1.000000_318.765378_L.png"
# hdf5_file_name = "4yoo1j1p65.h5"

# file_name = "0.637876_529822.070453_1057669646578.758545_0.461463_2195271249.769088_0.191704_1.000000_243.783366_L.png"
# hdf5_file_name = "hditjq8vx4.h5"


grrt_input_file = grrt_dir + file_name
ddpm_imput_file = ddpm_dir + file_name

origin_imput_file = origin_dir + hdf5_file_name
```

```python
FOV_UNITS = "muas"
# 加载HDF5文件
hfp = h5py.File(origin_imput_file,'r')
# 从文件中读取所需元数据
dx = hfp['header']['camera']['dx'][()]
dsource = hfp['header']['dsource'][()]
lunit = hfp['header']['units']['L_unit'][()]
scale = hfp['header']['scale'][()]
freq = hfp['header']['freqcgs'][()]
# 计算视场大小（以微弧秒为单位）
fov_muas = dx / dsource * lunit * 2.06265e11
extent = [ -fov_muas/2, fov_muas/2, -fov_muas/2, fov_muas/2 ]
imagep = np.copy(hfp['pol']).transpose((1,0,2))
I = imagep[:,:,0]
hfp.close()
flux = I.sum()*scale
print(I.sum()*scale)
c_cgs = 2.99792458e10
k_cgs = 1.38064852e-16
Tb = c_cgs**2/(2*freq**2*k_cgs) * I
```

```python
grrt_im = np.array(Image.open(grrt_input_file).convert('L'))
ddpm_im = np.array(Image.open(ddpm_imput_file).convert('L'))


nx, ny = grrt_im.shape
grrt_scale = flux/grrt_im.sum()
ddpm_scale = flux/ddpm_im.sum()

grrt_im_cgs = grrt_im*grrt_scale
ddpm_im_cgs = ddpm_im*ddpm_scale

grrt_im_Tb = c_cgs**2/(2*freq**2*k_cgs) * grrt_im_cgs
ddpm_im_Tb = c_cgs**2/(2*freq**2*k_cgs) * ddpm_im_cgs
```

```python
fig1, ax1 = plt.subplots(1,1,figsize=(3,3),dpi=150)
im = ax1.imshow(Tb, cmap='afmhot', vmin=0., vmax=Tb.max(), origin='lower')
ax1.axis('off')

# 创建水平colorbar
divider = make_axes_locatable(ax1)
cax = divider.append_axes("bottom", size="5%", pad=0.05)
cbar = plt.colorbar(im, cax=cax, orientation='horizontal')

# 设置刻度,除以10e10并只保留整数位置
vmax = Tb.max()
print(vmax/1e10)
ticks = np.arange(0, int(vmax/1e10)+1,1)  # 生成11个点以便找到整数位置
if len(ticks) > 6:
    ticks = ticks[::2]
cbar.set_ticks(ticks*1e10)
cbar.set_ticklabels([f'{int(t)}' for t in ticks])
plt.savefig(output_dir + file_name.replace(".png", "_grrt.png"), dpi=150)

plt.show()
plt.close()
```
